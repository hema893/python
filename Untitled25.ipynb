```python
Q2. Given are some following variables containing data:
(i) var1 = ‘ ‘
(ii) var2 = ‘[ DS , ML , Python]’
(iii) var3 = [ ‘DS’ , ’ML’ , ‘Python’ ]
(iv) var4 = 1.
```

```python
(i) var1 = ' '
👉 Data type: str (String)
Because it contains a space inside quotes.

(ii) var2 = '[ DS , ML , Python]'
👉 Data type: str (String)
Because it is enclosed in quotes, so it is treated as a string, not a list.

(iii) var3 = ['DS', 'ML', 'Python']
👉 Data type: list
Because elements are enclosed inside square brackets [ ].

(iv) var4 = 1.
👉 Data type: float
Because 1. represents a decimal number.
```
